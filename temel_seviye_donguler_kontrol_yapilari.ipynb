## 1. Hızlı Tekrar: Değişkenler ve Veri Yapıları

**Değişken**, veriyi saklamak için kullandığımız isimlendirilmiş bir alandır (bilgiyi tuttuğumuz kutu).

**İsimlendirme kuralları**
- Rakamla başlanmaz (`1_yas` hatalıdır).
- Boşluk içermez (`ogrenci yas` hatalıdır, `ogrenci_yas` doğrudur).
- Özel karakter içermez.
- Büyük-küçük harf duyarlıdır (`yas` ile `Yas` farklı değişkenlerdir).

**Temel veri tipleri**

| Tip | Anlamı | Örnek |
| :--- | :--- | :--- |
| `int` | Tam sayı | `35`, `-15` |
| `float` | Ondalıklı sayı | `3.14`, `99.99` |
| `str` | Metin | `"kaan"`, `'python'` |
| `bool` | Mantıksal değer (koşullarda kullanacağız) | `True`, `False` |

````python
yas = 35              # int
pi = 3.14             # float
isim = "kaan"         # str
ogrenci = True        # bool

print(type(yas))
print(type(pi))
print(type(isim))
print(type(ogrenci))
````

**Tip dönüşümü (casting):** `int()`, `float()`, `str()` fonksiyonlarıyla yapılır. `input()` her zaman `str` döndürür; sayı olarak kullanmak için çevirmeyi unutmayın. **f-string** ise metin içine değişken yerleştirmenin en pratik yoludur.

````python
x = "25"
print(int(x) + 5)       # 30
print(float(x))         # 25.0
print(str(35) + " yaş") # 35 yaş

kurulum_tarihi = 2023
print(f"Ucanble teknoloji {kurulum_tarihi} yılında kurulmuştur.")
````

**Veri yapıları (özet)**

| Yapı | Sözdizimi | Sıralı mı? | Değiştirilebilir mi? | Tekrar eden eleman |
| :--- | :--- | :---: | :---: | :--- |
| `list` | `[1, 2, 3]` | Evet | Evet | Olur |
| `tuple` | `(1, 2, 3)` | Evet | Hayır | Olur |
| `dict` | `{"anahtar": "değer"}` | Anahtarla erişilir | Evet | Anahtar tekrar edemez, değer edebilir |
| `set` | `{1, 2, 3}` | Hayır (indeks yok) | Evet | Olmaz |

Döngülerde en çok **liste, string, `range()` ve sözlük** üzerinde dolaşacağız.

````python
liste = [10, 20, 30]
tup = (10, 20, 30)
sozluk = {"isim": "ali", "yas": 25}
benzersiz = set([1, 2, 2, 3])

print(liste[0], liste[-1], liste[1:])   # indeksleme ve slicing
print(tup[1])
print(sozluk["isim"])
print(benzersiz)

liste.append(40)                        # liste değiştirilebilir
sozluk["not"] = 85                      # sözlüğe yeni anahtar
print(liste)
print(sozluk)
````

## 2. Kontrol Yapıları: `if` / `elif` / `else`

Kontrol yapıları, programın çalışma akışını **koşullara** bağlı olarak değiştirir. Koşul doğruysa (`True`) bir kod bloğu, yanlışsa başka bir kod bloğu çalışır.

```python
if kosul:
    yapilacak_islem
```

**Dikkat:** Koşulun sonundaki iki nokta `:` ve bloğun **girintisi (indentation)** zorunludur. Girinti unutulursa `IndentationError` alırsınız.

### `if`

````python
sayi = 10
if sayi > 0:                    # koşul: sayı 0'dan büyük mü?
    print("Sayı pozitiftir.")   # koşul doğruysa bu satır çalışır

# if sayi > 0:
# print("burası çalışmaz")      # IndentationError: girinti yok
````

### `if` - `else`

`else` bloğu, koşul **yanlış** olduğunda çalışır.

````python
sayi = -3

if sayi > 0:
    print("pozitif")
else:
    print("pozitif değil")
````

### `if` - `elif` - `else`

Koşullar yukarıdan aşağıya kontrol edilir. **İlk doğru koşulun bloğu çalışır, diğerleri kontrol edilmez.** Hiçbiri doğru değilse `else` çalışır.

````python
ogrenci_not = 72

if ogrenci_not > 85:
    print("A")
elif ogrenci_not > 70:
    print("B")      # 72 > 70 olduğu için burada durur
elif ogrenci_not > 50:
    print("C")      # kontrol edilmez bile
else:
    print("F")
````

### Karşılaştırma operatörleri

| Operatör | Anlamı |
| :---: | :--- |
| `==` | eşit mi? |
| `!=` | eşit değil mi? |
| `>` / `<` | büyük / küçük mü? |
| `>=` / `<=` | büyük eşit / küçük eşit mi? |

> `=` **atama** yapar (`x = 5`), `==` ise **karşılaştırır** (`x == 5`). İkisini karıştırmayın.

````python
a = 7
b = 3

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a <= b)   # False
print(a % 2 == 1)  # True -> a tek sayı
````

### Mantıksal operatörler: `and`, `or`, `not`

Birden fazla koşulu birleştirir.

| Operatör | Ne zaman `True`? |
| :---: | :--- |
| `and` | **İki koşul da** doğruysa |
| `or` | **En az biri** doğruysa |
| `not` | Koşulu tersine çevirir |

````python
yas = 20
ogrenci = True

print("and:", yas < 25 and ogrenci == True)   # ikisi de True -> True
print("or :", yas > 25 or ogrenci == True)    # biri True -> True
print("not:", not ogrenci)                    # False

if yas < 25 and ogrenci:
    print("Öğrenci indirimi uygula")
````

### `in` ve `not in` ile üyelik kontrolü

Bir elemanın liste, string veya başka bir koleksiyonda olup olmadığını kontrol eder.

````python
meyveler = ["elma", "armut", "muz"]
urun = "kivi"          # urun = input("Bir meyve girin: ") ile kullanıcıdan da alınabilir

if urun in meyveler:
    print("Stokta var")
else:
    print("Stokta yok")

print("elma" in meyveler)        # True
print("kivi" not in meyveler)    # True
print("y" in "python")           # True (string içinde karakter arama)
````

## 3. `for` Döngüsü

**Döngü (loop)**, bir işlemi tekrar tekrar yapmamızı sağlar. Örneğin bir listedeki tüm elemanları yazdırmak gibi.

```python
for degisken in koleksiyon:
    yapilacak_islem
```

- `degisken`: her turda (iterasyonda) değişen geçici isim
- `koleksiyon`: liste, tuple, string, sözlük, `range()` gibi dolaşılabilir yapılar

Döngü olmadan `sayilar[0] + 5`, `sayilar[1] + 5`, `sayilar[2] + 5` ... diye tek tek yazmak gerekirdi; eleman sayısı arttıkça bu yöntem sürdürülemez.

````python
sayilar = [10, 20, 30]

for sayi in sayilar:
    print(sayi + 5)
````

### `range()` fonksiyonu

Sayı dizisi üretir. **Bitiş değeri dahil değildir.**

| Kullanım | Ürettiği sayılar |
| :--- | :--- |
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 7)` | 1, 2, 3, 4, 5, 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (adım = 2) |
| `range(10, 0, -1)` | 10, 9, 8, ..., 1 (geri sayım) |

````python
print(list(range(5)))
print(list(range(1, 7)))
print(list(range(0, 10, 2)))
print(list(range(10, 0, -1)))

for i in range(1, 4):
    print("i:", i)
````

### `for` ile toplama ve iz sürme

Döngü içinde bir değişkeni her turda güncelleyerek toplam hesaplayabiliriz. Başlangıç değerini (`toplam = 0`) döngüden **önce** vermeyi unutmayın.

| Tur | `s` | `toplam` (işlemden önce) | `toplam` (işlemden sonra) |
| :---: | :---: | :---: | :---: |
| 1 | 10 | 0 | 10 |
| 2 | 20 | 10 | 30 |
| 3 | 30 | 30 | 60 |

````python
sayilar = [10, 20, 30]

toplam = 0
for s in sayilar:
    toplam = toplam + s          # kısaca: toplam += s
    print(f"s: {s}, toplam: {toplam}")

print("Sonuç:", toplam)          # 60
````

### `for` + `if`

Döngü içinde koşul kullanarak elemanları filtreleyebiliriz.

````python
sayilar = [1, 2, 3, 4, 5, 6]

for sayi in sayilar:
    if sayi % 2 == 0:
        print(f"Çift: {sayi}")
````

### String üzerinde dolaşma

Bir string, karakter dizisidir; `for` ile harf harf dolaşılabilir.

````python
kelime = "ucanble"

for harf in kelime:
    print(harf)
````

### Sözlük üzerinde dolaşma

- Doğrudan dolaşmak **anahtarları** verir.
- `.items()` hem anahtarı hem değeri verir.

````python
ogrenci = {"isim": "ali", "yas": 25, "bolum": "bilgisayar"}

for anahtar in ogrenci:
    print(anahtar)

print("-" * 20)

for anahtar, deger in ogrenci.items():
    print(f"{anahtar}: {deger}")
````

### `enumerate()` ile indeks ve eleman birlikte

Hem sırayı (indeks) hem elemanı almak istediğimizde kullanılır.

````python
meyveler = ["elma", "armut", "muz"]

for indeks, meyve in enumerate(meyveler):
    print(indeks, meyve)
````

### Sık kullanılan döngü kalıpları

| Kalıp | Başlangıç | Döngü içinde |
| :--- | :--- | :--- |
| Toplam | `toplam = 0` | `toplam += eleman` |
| Sayaç | `adet = 0` | `adet += 1` |
| En büyük | `en_buyuk = liste[0]` | `if eleman > en_buyuk: en_buyuk = eleman` |

````python
notlar = [70, 85, 90, 65, 100]

adet = 0
toplam = 0
en_buyuk = notlar[0]

for n in notlar:
    adet += 1
    toplam += n
    if n > en_buyuk:
        en_buyuk = n

print("Adet:", adet)
print("Toplam:", toplam)
print("Ortalama:", toplam / adet)
print("En büyük:", en_buyuk)
````

## 4. `while` Döngüsü

`while`, **koşul doğru olduğu sürece** çalışan bir döngüdür.

```python
while kosul:
    yapilacak_islem
```

**`if` ile farkı:** `if` koşulu yalnızca **bir kez** kontrol eder. `while` ise koşulu **her turda** yeniden kontrol eder.

**`for` ile farkı:** Kaç tur döneceğini önceden biliyorsak (liste, `range`) `for`; ne zaman biteceği bir koşula bağlıysa (kullanıcı `q` yazana kadar gibi) `while` kullanırız.

````python
i = 0
while i < 5:        # i 5'ten küçük olduğu sürece
    print(i)
    i = i + 1       # sayacı artırmak ŞART, yoksa döngü hiç bitmez

print(f"i: {i}")
````

**İz sürme tablosu**

| Tur | `i` (tur başında) | `i < 5` | Ekrana yazılan | `i` (tur sonunda) |
| :---: | :---: | :---: | :---: | :---: |
| 1 | 0 | True | 0 | 1 |
| 2 | 1 | True | 1 | 2 |
| 3 | 2 | True | 2 | 3 |
| 4 | 3 | True | 3 | 4 |
| 5 | 4 | True | 4 | 5 |
| 6. kontrol | 5 | **False** | - | Döngü biter |

6. tur **yoktur**: `i` 5 olduğunda `5 < 5` koşulu yanlış olur ve döngüye girilmez. Döngü bittiğinde `i` değeri 5'tir.

> **Sonsuz döngü uyarısı:** Koşulu `False` yapacak bir güncelleme (`i += 1` gibi) yoksa döngü sonsuza kadar çalışır. Jupyter'da takılırsanız **Kernel → Interrupt** ile durdurabilirsiniz.

````python
# sayaç mantığı
sayac = 1
while sayac <= 3:
    print("merhaba")
    sayac += 1
````

````python
# while + if
i = 0
while i < 10:
    if i % 2 == 0:
        print(f"çift: {i}")
    i += 1
````

### Kullanıcı kontrollü `while`

Kullanıcı `q` yazana kadar giriş almak klasik bir `while` örneğidir. Aşağıda `input()` yerine hazır bir liste kullanıyoruz ki notebook kendi kendine çalışsın.

```python
# Gerçek kullanım:
giris = ""
while giris != "q":
    giris = input("Çıkmak için q yazın: ")
    print(f"Kullanıcı mesajı: {giris}")
```

````python
girisler = ["merhaba", "python", "q", "bu hiç okunmaz"]   # kullanıcı girişlerini simüle ediyoruz
sira = 0

giris = ""
while giris != "q":
    giris = girisler[sira]        # giris = input("Çıkmak için q yazın: ")
    sira += 1
    print(f"Kullanıcı mesajı: {giris}")

print("Çıkış yapıldı")
````

Dikkat: `q` yazıldığında da o turun kodu (yukarıdaki `print`) sonuna kadar çalışır; koşul ancak bir sonraki turun başında kontrol edilir. Bu yüzden çıktıda `q` mesajı da görünür.

### Chatbot örneği

Gerçek bir chatbot da aynı mantıkla çalışır: kullanıcıdan mesaj al, botun cevabını yazdır, `q` gelene kadar tekrar et.

````python
girisler = ["selam", "nasılsın", "q"]
sira = 0

giris = ""
while giris != "q":
    giris = girisler[sira]
    sira += 1
    print(f"Kullanıcı mesajı: {giris}")

    # 1. chatbota soruyu gönder
    # 2. chatbot cevabı return eder
    # 3. cevabı ekrana yazdır
    print("chatbot: merhaba")
````

## 5. `break`, `continue` ve `pass`

| Komut | Ne yapar? |
| :--- | :--- |
| `break` | Döngüyü **tamamen durdurur**; koşul sağlanınca döngüden çıkılır ve bir daha devam edilmez. |
| `continue` | **O anki turu atlar**, döngü bir sonraki turla devam eder. |
| `pass` | **Hiçbir şey yapmaz.** Henüz kod yazmadığımız yeri boş bırakmak için kullanılır; program hata vermez. |

### `break`

````python
for i in range(10):
    if i == 5:
        break        # i 5 olunca döngü tamamen biter
    print(i)         # 0, 1, 2, 3, 4
````

````python
# break ile arama: ilk bulduğunda dur
sayilar = [4, 9, 12, 17, 20, 25]

for s in sayilar:
    if s > 15:
        print("15'ten büyük ilk sayı:", s)
        break        # geri kalanına bakmaya gerek yok
````

### `continue`

````python
for i in range(10):
    if i == 5:
        continue     # sadece 5. tur atlanır, döngü devam eder
    print(i)         # 0, 1, 2, 3, 4, 6, 7, 8, 9
````

### `pass`

````python
if True:
    # burayı sonra doldur
    pass

for i in range(3):
    if i == 1:
        pass         # TODO: i 1 ise buraya bir şeyler yapılacak
    print(i)         # pass hiçbir şeyi atlamaz: 0, 1, 2
````

> **`pass` ile `continue` farkı:** `pass` sadece boşluk doldurur, altındaki kodlar çalışmaya devam eder. `continue` ise turun geri kalanını atlar.

### `while` içinde `break`

`while True:` ile "sonsuz" bir döngü kurup, uygun anda `break` ile çıkmak yaygın bir kalıptır.

````python
girisler = ["a", "b", "q", "c"]
sira = 0

while True:                    # koşul her zaman doğru
    giris = girisler[sira]
    sira += 1
    if giris == "q":
        break                  # çıkışı biz kontrol ediyoruz
    print("Girdiniz:", giris)

print("Çıkış yapıldı")
````

## 6. İç İçe (Nested) Yapılar

Yapıların birbirinin içinde kullanılmasıdır: `for` içinde `if`, `if` içinde `if`, `for` içinde `for`, `if` içinde `while` gibi.

### `for` içinde `if`

````python
for i in range(5):
    if i % 2 == 0:
        print(i)
````

### `if` içinde `if`

````python
yas = 20
ogrenci = True

if yas < 25:
    if ogrenci:
        print("Öğrenci indirimi")
````

### `for` içinde `for`

Dıştaki döngünün **her turu için** içteki döngü **baştan sona** çalışır.

| Dış tur (`i`) | İç turlar (`j`) |
| :---: | :--- |
| 0 | 0, 1 |
| 1 | 0, 1 |
| 2 | 0, 1 |

````python
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
````

`print()` varsayılan olarak satır sonuna geçer. `end=""` parametresiyle aynı satırda yazmaya devam edebiliriz. Aşağıda iç içe döngüyle yıldız üçgeni çiziyoruz:

````python
for satir in range(1, 5):
    for _ in range(satir):     # her satırda, satır numarası kadar yıldız
        print("*", end="")
    print()                    # satırı bitir, alt satıra geç
````

### İç içe döngüde `break`

`break` ve `continue` **sadece bulundukları (en içteki) döngüyü** etkiler.

````python
for i in range(3):
    for j in range(3):
        if j == 2:
            break              # sadece içteki döngüyü keser
        print(i, j)
````

## 7. Özet

| Yapı | Ne zaman kullanılır? | Temel yazım |
| :--- | :--- | :--- |
| `if / elif / else` | Koşula göre farklı kod çalıştırmak | `if kosul:` |
| `for` | Koleksiyon veya `range` üzerinde, tur sayısı belliyken | `for x in koleksiyon:` |
| `while` | Tur sayısı belli değil, bir koşula bağlı | `while kosul:` |
| `break` | Döngüyü tamamen bitirmek | `break` |
| `continue` | Sadece o turu atlamak | `continue` |
| `pass` | Boş bırakılan blok (hata vermesin) | `pass` |
| Nested | Yapıları iç içe kullanmak | girintiye dikkat |

**Hatırlatmalar**
- `:` ve girinti zorunludur.
- `while` içinde sayacı/koşulu güncellemeyi unutmayın (sonsuz döngü).
- `range(a, b)` bitiş değeri `b`'yi **içermez**.
- `=` atama, `==` karşılaştırmadır.

## Soru 1 (IF)

Kullanıcıdan bir sayı alın. Sayı pozitifse "Pozitif", negatifse "Negatif", sıfırsa "Sıfır" yazdırın.

````python
sayi = -5      # sayi = int(input("Bir sayı girin: "))

if sayi > 0:
    print("Pozitif")
elif sayi < 0:
    print("Negatif")
else:
    print("Sıfır")
````

## Soru 2 (FOR)

1'den 10'a kadar (10 dahil) sayıları yazdırın. Ayrıca bu sayıların toplamını hesaplayıp ekrana yazdırın.

````python
toplam = 0

for i in range(1, 11):
    print(i)
    toplam += i

print("Toplam:", toplam)
````

## Soru 3 (WHILE)

Kullanıcıdan "q" yazana kadar sürekli giriş alın. Kullanıcı her giriş yaptığında "Girdiniz: ..." şeklinde ekrana yazdırın. Kullanıcı "q" yazarsa döngü bitsin ve "Çıkış yapıldı" yazsın.

````python
girisler = ["merhaba", "python", "q"]   # kullanıcı girişlerini simüle ediyoruz
sira = 0

giris = ""
while giris != "q":
    giris = girisler[sira]              # giris = input("Bir şey yazın (çıkmak için q): ")
    sira += 1
    if giris != "q":
        print(f"Girdiniz: {giris}")

print("Çıkış yapıldı")
````

## Soru 4 (NESTED)

1'den 20'ye kadar sayıları dolaşın. Sayı çiftse "Çift", tekse "Tek" yazdırın. Ayrıca sayı 10'dan büyükse yanına "Büyük", değilse "Küçük/Eşit" yazdırın.

Örnek çıktı: `12 -> Çift - Büyük`

````python
for i in range(1, 21):
    # İlk kontrol: çift mi tek mi?
    if i % 2 == 0:
        tur = "Çift"
    else:
        tur = "Tek"

    # İkinci kontrol: 10'dan büyük mü?
    if i > 10:
        boyut = "Büyük"
    else:
        boyut = "Küçük/Eşit"

    print(f"{i} -> {tur} - {boyut}")
````
