# Temel Seviye Python: Değişkenler ve Veri Yapıları

Bu ders notu, Python programlama dilinde değişken kavramı, temel veri tipleri (`int`, `float`, `str`), veri tipi kontrolleri, tip dönüşümleri ve gelişmiş veri yapıları (`list`, `tuple`, `dict`, `set`) konularını kapsamaktadır.

## 1. Değişken Kavramı ve İsimlendirme Kuralları

- **Değişken**: Veriyi saklamak için kullandığımız isimlendirilmiş bir alandır (bilgiyi tuttuğumuz kutu).
- **İsimlendirme Kuralları**:
  - Rakamla başlamaz (Örn: `1_var` **hatalıdır**).
  - Boşluk içermez (Örn: `kaan hoca` **hatalıdır**).
  - Özel karakter içermez.
  - Büyük-küçük harf duyarlıdır (`degisken` ile `Degisken` aynı şey değildir).

## 2. Temel Veri Tipleri

### Integer (`int`) - Tam Sayılar
Tam sayıları temsil eden veri tipidir.

```python
# Örnek tanımlamalar
yas = 35
ogrenci_sayisi = 55000
sicaklik = -15 

print(f"Yaş: {yas}")
print(f"Öğrenci Sayısı: {ogrenci_sayisi}")
print(f"Sıcaklık: {sicaklik}")
```

```python
# Temel Matematiksel İşlemler
a = 10
b = 5 

print("Toplam:", a + b)
print("Çarpma:", a * b)
print("Çıkarma:", a - b)
print("Bölme:", a / b)
```

```python
# Gerçek Hayat Örneği: Ürün Fiyatı Hesaplama
urun_sayisi = 8 
birim_fiyat = 10 

toplam_tutar = urun_sayisi * birim_fiyat
print(f"Toplam Tutar: {toplam_tutar} TL")
```

### Float - Ondalıklı Sayılar
Ondalıklı sayıları temsil eden veri tipidir.

```python
pi = 3.14
sicaklik_ondalik = 35.5
urun_fiyati = 99.99

print("Pi Sayısı:", pi)
print("Sıcaklık:", sicaklik_ondalik)
print("Ürün Fiyatı:", urun_fiyati)
```

```python
# Bilgisayar matematiğinde ondalık hassasiyeti ve yuvarlama (round)
sonuc = 0.1 + 0.2
print("Normal Toplam:", sonuc)  # 0.30000000000000004

sonuc_yuvarlanmis = round(sonuc, 2)
print("Yuvarlanmış Sonuç:", sonuc_yuvarlanmis)
```

### String (`str`) - Metinsel Veriler
Karakter dizilerini temsil eder. Tek tırnak veya çift tırnak ile tanımlanabilir.

```python
isim = "kaan"
sirket = 'ucanble'

bilgi = "kaan hocanın şirketinin ismi ucanble teknoloji"
print(bilgi)

# String Birleştirme (Concatenation)
bilgi2 = isim + " hocanın şirketinin ismi " + sirket + " teknoloji"
print(bilgi2)
```

```python
# String ve Sayı Birleştirme & f-string Kullanımı
yas = 35
kurulum_tarihi = 2023

print(f"Ucanble Teknoloji {kurulum_tarihi} yılında kurulmuştur.")

accuracy = 95
print(f"Karar ağacı accuracy: {accuracy} %")
```

```python
# String İndeksleme ve Metotlar
kelime = "python"
print("İlk harf (0. indeks):", kelime[0])
print("4. indeks:", kelime[3])

metin = "PythoN"
print("Küçük harf:", metin.lower())
print("Uzunluk (len):", len(metin))
print("Karakter Değiştirme (replace):", metin.replace("o", "O"))
```

## 3. Veri Tipi Kontrolü ve Tip Dönüşümleri (Casting)

- `type()` fonksiyonu ile veri tipi öğrenilir.
- `int()`, `float()`, `str()` fonksiyonları ile tip dönüşümleri yapılır.

```python
x = 10
print(type(x))  # <class 'int'>

x = "10"
print(type(x))  # <class 'str'>

# Tip Dönüşümleri
x_str = "25"
print("Int dönüşümü:", type(int(x_str)))
print("Float dönüşümü:", type(float(x_str)))

sayi = 35
print("String dönüşümü:", type(str(sayi)))
```

## 4. Gelişmiş Veri Yapıları

### Listeler (`list`)
- Köşeli parantez `[]` ile tanımlanır.
- Sıralıdır, değiştirilebilir (mutable), farklı veri tiplerini aynı anda barındırabilir.

```python
sayilar = [1, 2, 3, 4, 5, 6]
isimler = ["kaan", "can", "ucanble", "yılmaz"]
karisik = ["kaan", 1, "can", "ucanble", 55, 65.5]

print("Karışık Liste:", karisik)

meyveler = ["elma", "muz", "kivi"]
print("0. indeks:", meyveler[0])
print("Son eleman (-1):", meyveler[-1])
print("Liste Uzunluğu:", len(meyveler))
```

```python
# Slicing (Dilimleme) [başlangıç:bitiş]
sayilar = [10, 20, 30, 40, 50, 60]
print("1. indexten 4. indexe:", sayilar[1:4])
print("İlk 3 eleman:", sayilar[:3])
print("2. indexten sonrasına kadar:", sayilar[2:])
```

```python
# Liste Metotları
sayilar = [1, 2, 3]
sayilar.append(4)
print("Append sonrası:", sayilar)

sayilar.insert(1, 100)
print("Insert sonrası:", sayilar)

sayilar.remove(100)
print("Remove sonrası:", sayilar)

sayilar.pop()
print("Pop sonrası:", sayilar)

sayilar[0] = 999
print("Güncelleme sonrası:", sayilar)
```

### Tuple (`tuple`)
- Normal parantez `()` ile tanımlanır.
- Sıralıdır ancak **değiştirilemez** (`immutable`).

```python
koordinat = (10, 20)
renkler = ("kırmızı", "mavi", "yeşil")

t = (10, 20, 30, 40)
print("1. indeks:", t[1])
print("Slicing (1:3):", t[1:3])

# Tek elemanlı tuple tanımlama
dogru_tek_elemanli = (5,)
print("Tip:", type(dogru_tek_elemanli))

# Tuple Unpacking
x, y = koordinat
print(f"x: {x}, y: {y}")

print("20 elemanının sayısı (count):", t.count(20))
print("30 elemanının indeksi (index):", t.index(30))
```

### Sözlük (`dict` - Dictionary)
- Süslü parantez `{}` ile anahtar-değer (`key-value`) mantığıyla saklanır.

```python
ogrenci = {
    "isim": "ali", 
    "yas": 25,
    "bolum": "bilgisayar"
}

print("Öğrenci Sözlüğü:", ogrenci)
print("İsim:", ogrenci["isim"])

ogrenci["not"] = 85
ogrenci["yas"] = 26
print("Güncel Sözlük:", ogrenci)

del ogrenci["bolum"]
print("Silme Sonrası:", ogrenci)

print("Keys:", ogrenci.keys())
print("Values:", ogrenci.values())
print("Items:", ogrenci.items())
```

### Set (`set` - Küme)
- Süslü parantez `{}` ile tanımlanır.
- Sırasızdır, indekslenemez ve **aynı elemandan birden fazla barındırmaz** (benzersiz).

```python
sayilar = {1, 2, 2, 3, 3, 3, 4}
print("Set Hali:", sayilar)

liste = [1, 2, 2, 3, 4, 4]
benzersiz = set(liste)
print("Listeden Set'e:", benzersiz)

sayilar.add(5)
print("Add(5):", sayilar)
sayilar.remove(2)
print("Remove(2):", sayilar)

a = {1, 2, 3}
b = {3, 4, 5}
print("Birleşim (Union):", a.union(b))
print("Kesişim (Intersection):", a.intersection(b))
print("Fark (Difference):", a.difference(b))
```

## 5. Özet: Veri Yapılarının Karşılaştırılması

| Veri Yapısı | Sıralı mı? | Değiştirilebilir mi? | Tekrar Eden Eleman? | Sözdizimi | Temel Kullanım Amacı |
| :--- | :---: | :---: | :---: | :---: | :--- |
| **List** | Evet | Evet | Evet | `[1, 2, 3]` | Sıralı ve güncellenebilir koleksiyonlar |
| **Tuple** | Evet | Hayır | Evet | `(1, 2, 3)` | Sabit kalması gereken güvenli veriler |
| **Dictionary**| - | Evet | Değerler: Evet (Key: Hayır) | `{"key": "value"}` | Etiketli, anahtar-değer ilişkili veriler |
| **Set** | Hayır | Evet | Hayır | `{1, 2, 3}` | Benzersiz eleman yönetimi ve küme işlemleri |
