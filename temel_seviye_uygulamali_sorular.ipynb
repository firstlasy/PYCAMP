## Uygulama 1: Maaş ve Zam Hesaplama
Bir çalışanın aylık maaşı 28500 TL'dir. Bu çalışana %17.5 oranında zam yapılacaktır. Yeni maaşı hesaplayıp ekrana yazdıran Python kodunu yazınız.

```python
# Buraya kodunuzu yazın:
maas = 28500
zam_orani = 17.5
yeni_maas = maas + (maas * zam_orani / 100)
print("Yeni Maaş:", yeni_maas)
```

## Uygulama 2: Daire Alanı Hesaplama
Yarıçapı `r = 7` olan bir dairenin alanını hesaplayınız (Formül: $\pi \times r^2$, $\pi = 3.14$). Sonucu float olarak ekrana yazdırın.

```python
# Buraya kodunuzu yazın:
pi = 3.14
r = 7
alan = pi * (r ** 2)
print("Dairenin Alanı:", alan)
```

## Uygulama 3: Saniye Dönüştürücü
Kullanıcıdan (veya sabit bir değişkenden) alınan 5000 saniyenin kaç dakika ve kaç saniye olduğunu hesaplayan Python kodunu yazınız.

## Uygulama 4: Metin Manipülasyonu ve Slicing
`cumle = "Yapay Zeka ve Veri Bilimi Akademisi"` metninden slicing yaparak sadece **"Veri Bilimi"** kısmını çekip büyük harfle yazdırınız.

## Uygulama 5: Kelime ve Karakter Sayısı
`metin = "programlama dili python"` ifadesinin toplam karakter uzunluğunu (`len`) bulan ve kelimeleri boşluklardan bölerek (`split`) toplam kelime sayısını ekrana yazdıran kodu yazınız.

```python
# Buraya kodunuzu yazın:
metin = "programlama dili python"
karakter_sayisi = len(metin)
kelimeler = metin.split()
kelime_sayisi = len(kelimeler)
print("Karakter Sayısı:", karakter_sayisi)
print("Kelime Sayısı:", kelime_sayisi)
```

## Uygulama 7: Liste Elemanlarını Filtreleme ve Sıralama
`notlar = [45, 85, 60, 90, 30, 75, 100]` listesini küçükten büyüğe sıralayın (`sort`), ardından en yüksek ve en düşük notu ekrana yazdırın.

```python
# Buraya kodunuzu yazın:
notlar = [45, 85, 60, 90, 30, 75, 100]
notlar.sort()
print("Sıralı Liste:", notlar)
print("En Düşük Not:", notlar[0])
print(
    "En Yüksek Not:", notlar[-1]
)  # veya max(notlar), min(notlar) kullanılabilir
```

## Uygulama 8: Liste Elemanlarının Ortalaması
`sayilar = [12, 45, 67, 23, 89, 34]` listesindeki tüm sayıların toplamını `sum()` ve eleman sayısını `len()` kullanarak aritmetik ortalamasını hesaplayın.

```python
# Buraya kodunuzu yazın:
sayilar = [12, 45, 67, 23, 89, 34]
ortalama = sum(sayilar) / len(sayilar)
print("Aritmetik Ortalama:", round(ortalama, 2))
```

## Uygulama 9: Alışveriş Sepeti (Liste Manipülasyonu)
Boş bir `sepet = []` listesi ile başlayın. Sırasıyla sepete `"Elma"`, `"Armut"`, `"Muz"` ekleyin (`append`). Ardından `"Armut"` elemanını listeden silin (`remove`) ve son sepet halini yazdırın.

```python
# Buraya kodunuzu yazın:
sepet = []
sepet.append("Elma")
sepet.append("Armut")
sepet.append("Muz")
sepet.remove("Armut")
print("Güncel Sepet:", sepet)
```

## Uygulama 10: İki Listenin Birleştirilmesi ve Genişletilmesi
`grup1 = ["Ali", "Ayşe"]` ve `grup2 = ["Mehmet", "Can"]` listelerini tek bir liste halinde birleştirin (`extend` veya `+` operatörü ile) ve toplam eleman sayısını yazdırın.

```python
# Buraya kodunuzu yazın:
grup1 = ["Ali", "Ayşe"]
grup2 = ["Mehmet", "Can"]
tum_liste = grup1 + grup2
print(
    "Birleştirilmiş Liste:", tum_liste, "Toplam Kişi:", len(tum_liste)
)
```

## Uygulama 11: Tuple Unpacking Uygulaması
`gps_konumu = (37.1234, 28.5678, "Muğla")` tuple verisi veriliyor. Bu veriyi unpacking yaparak `enlem`, `boylam`, `sehir` değişkenlerine atayın ve ekrana düzgün bir cümle olarak yazdırın.

```python
# Buraya kodunuzu yazın:
gps_konumu = (37.1234, 28.5678, "Muğla")
enlem, boylam, sehir = gps_konumu
print(
    f"{sehir} şehrinin koordinatları -> Enlem: {enlem}, Boylam: {boylam}"
)
```

## Uygulama 12: Tuple İçinde Arama (count ve index)
`harfler = ('a', 'b', 'c', 'a', 'd', 'a', 'e')` tuple içinde `'a'` karakterinin kaç kez geçtiğini (`count`) ve ilk geçtiği indeksi (`index`) bulan kodu yazınız.

```python
# Buraya kodunuzu yazın:
harfler = ('a', 'b', 'c', 'a', 'd', 'a', 'e')
adet = harfler.count('a')
ilk_indeks = harfler.index('a')
print("'a' harfi adedi:", adet)
print("İlk geçtiği indeks:", ilk_indeks)
```

## Uygulama 13: E-Ticaret Ürün Sözlüğü
Anahtarları `"urun_adi"`, `"fiyat"`, `"stok"`, `"kategori"` olan bir `urun` sözlüğü oluşturun. Değerleri sırasıyla `"Laptop"`, `35000`, `15`, `"Elektronik"` olsun. Sözlüğe `"indirim_orani": 10` anahtarını ekleyin.

## Uygulama 14: Sözlük Güncelleme ve Fiyat Hesaplama
Uygulama 13'teki ürünün fiyatına %20 zam yapın (yeni fiyatı güncelleyin) ve stok miktarını 5 azaltın. Son sözlüğü yazdırın.

## Uygulama 15: Sözlükte Güvenli Değer Okuma (get metodu)
Bir öğrenci sözlüğü tanımlayın: `ogrenci = {"isim": "Zeynep", "numara": 452}`. Bu sözlükten `.get()` metodunu kullanarak `"not"` anahtarını sorgulayın (hata almadan `"Bulunamadı"` mesajı döndürmesini sağlayın).

```python
# Buraya kodunuzu yazın:
ogrenci = {"isim": "Zeynep", "numara": 452}
ogrenci_not = ogrenci.get("not", "Not bilgisi bulunamadı")
print("Sorgu Sonucu:", ogrenci_not)
```

## Uygulama 16: Liste ve Set Karşılaştırması (Tekrar Edenleri Temizleme)
`ogrenci_idleri = [101, 102, 101, 103, 104, 102, 105, 103]` listesinde tekrar eden ID'ler bulunmaktadır. Bu listeyi `set` veri yapısına dönüştürerek benzersiz ID listesini elde edin ve tekrar eden kayıt sayısını bulun.

```python
# Buraya kodunuzu yazın:
ogrenci_idleri = [101, 102, 101, 103, 104, 102, 105, 103]
benzersiz_idler = set(ogrenci_idleri)
tekrar_eden_sayisi = len(ogrenci_idleri) - len(benzersiz_idler)
print("Benzersiz ID'ler:", benzersiz_idler)
print("Tekrar Eden Kayıt Sayısı:", tekrar_eden_sayisi)
```

## Uygulama 17: Küme İşlemleri (Ortak Ders Seçimleri)
`matematik_ogrencileri = {"Ali", "Ayşe", "Fatma", "Can"}` ve `fizik_ogrencileri = {"Ayşe", "Can", "Mehmet", "Kemal"}` kümeleri veriliyor. Her iki derse de kayıtlı olan öğrencileri (`intersection`) bulan kodu yazın.

```python
# Buraya kodunuzu yazın:
matematik_ogrencileri = {"Ali", "Ayşe", "Fatma", "Can"}
fizik_ogrencileri = {"Ayşe", "Can", "Mehmet", "Kemal"}
her_iki_ders = matematik_ogrencileri.intersection(fizik_ogrencileri)
print("Her İki Derse Alan Öğrenciler:", her_iki_ders)
```

## Uygulama 18: Küme Fark İşlemi
Soru 17'deki verileri kullanarak **sadece** Matematik alan ve Fizik almayan öğrencileri bulan (`difference`) Python kodunu yazınız.

```python
# Buraya kodunuzu yazın:
sadece_matematik = matematik_ogrencileri.difference(fizik_ogrencileri)
print("Sadece Matematik Alanlar:", sadece_matematik)
```

Kullanıcıdan input alındığını simüle eden `giris = "125.75"` string ifadesini önce `float` türüne, ardından tam sayı kısmını almak için `int` türüne dönüştürün ve aradaki farkları ekrana yazdırın.

## Uygulama 20: Kapsamlı Veri Analizi Prototipi
Bir öğrencinin aldığı dersler ve puanları sözlük olarak verilmiştir:
`ogrenci_notlari = {"Matematik": 85, "Fizik": 70, "Kimya": 90, "Türkçe": 78}`
Bu sözlükteki tüm notların ortalamasını hesaplayıp ekrana yazdıran Python kodunu yazınız.

```python
# Buraya kodunuzu yazın:
ogrenci_notlari = {"Matematik": 85, "Fizik": 70, "Kimya": 90, "Türkçe": 78}
notlar_listesi = list(ogrenci_notlari.values())
ortalama = sum(notlar_listesi) / len(notlar_listesi)
print("Öğrencinin Ders Notları Ortalaması:", round(ortalama, 2))
```
